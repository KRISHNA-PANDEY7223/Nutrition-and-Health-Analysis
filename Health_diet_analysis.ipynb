```python
import seaborn as sns
import matplotlib.pyplot as plt

G_al = pd.crosstab(
    df["Gender"],
    df["Activity_Level"],
    normalize="index"
) * 100

activity_order = [
    "Sedentary",
    "Lightly Active",
    "Moderately Active",
    "Very Active",
    "Athlete"
]

G_al = G_al.reindex(columns=activity_order)

G_al.plot(kind="bar", stacked=True, figsize=(8, 5))

plt.title("Gender-wise Activity Level Distribution")
plt.xlabel("Gender")
plt.ylabel("Percentage")
plt.legend(title="Activity Level")
plt.tight_layout()
plt.show()
```

```python
sns.countplot(
    data=df,
    x="Gender",
    hue="Activity_Level",
    hue_order=[
        "Sedentary",
        "Lightly Active",
        "Moderately Active",
        "Very Active",
        "Athlete"
    ]
)

plt.title("Gender-wise Activity Level")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

gender_cal = df.groupby("Gender")[["Daily_Calorie_Requirement",
                                   "Daily_Calorie_Consumed"]].mean().reset_index()

gender_cal = gender_cal.melt(
    id_vars="Gender",
    var_name="Metric",
    value_name="Calories"
)

sns.barplot(
    data=gender_cal,
    x="Gender",
    y="Calories",
    hue="Metric"
)

plt.title("Average Calorie Intake vs Calories Consumed by Gender")
plt.show()
```

```python
g_hs=pd.crosstab(
    df["Gender"],
    df["Health_Status"],
    normalize='index'
)*100

g_hs=g_hs[["Underweight","Healthy","Overweight","Obese"]]

sns.countplot(data=g_hs,x=df["Gender"],hue=df["Health_Status"])
plt.title("Gender wise Health Report")
plt.Figure(figsize=(10,6))
plt.tight_layout()
plt.show()
```

```python
sns.regplot(data=df,x=df["Weight_kg"],y=df["BMI"],line_kws={"color": "yellow", "linewidth": 2,"linestyle":"--","alpha":0.6})
plt.title("weight wise bmi")
plt.show()
```

```python
sns.lineplot(data=df,x=df["BMI"],y= df["Health_Status"],hue_order=["Underweight","Healthy","Overweight","Obese"],hue=df["Health_Status"])
plt.title("BMI vs Health_Status")
plt.show()
```

```python
nutrients = (
    df.groupby("Diet_Type")[["Fat_Intake_g", "Protein_Intake_g"]].mean().reset_index()
)

nutrients_long = nutrients.melt(
    id_vars="Diet_Type",
    var_name="Nutrient",
    value_name="Average_Intake"
)

sns.barplot(
    data=nutrients_long,
    x="Diet_Type",
    y="Average_Intake",
    hue="Nutrient"
)

plt.title("Average Fat and Protein Intake by Diet Type")
plt.xticks(rotation=45)
plt.show()
```
